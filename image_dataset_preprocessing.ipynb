```python
# ============================================================
# CELL 1: INSTALL AND IMPORT LIBRARIES
# ============================================================

# Install KaggleHub
!pip install -q kagglehub                    #-q unneccary installations messages are reduced

# Import required libraries
import os                               #helps in workout of files,folders,file paths
import cv2                              #reading image ,resizing,color conversion,filtering  etc..,
import numpy as np                      #numerical data
import matplotlib.pyplot as plt          # graphs
import kagglehub                       #data set taken from kaggle

from PIL import Image               #another way of opening and saving images

print("BEFORE:")
print("Required libraries are not imported.")

print("\nAFTER:")
print("✓ kagglehub imported")
print("✓ OpenCV imported")
print("✓ NumPy imported")
print("✓ Matplotlib imported")
print("✓ PIL imported")
print("\nAll required libraries are ready!")
```

```python
# ============================================================
# CELL 2: DOWNLOAD KAGGLE DATASET
# ============================================================

dataset_id = "mahmoudalforawi/igneous-metamorphic-sedimentary-rocks-and-minerals"

print("BEFORE:")
print("Dataset has not been downloaded.")

# Download the dataset
dataset_path = kagglehub.dataset_download(dataset_id)                     #it returns the url location

print("\nAFTER:")
print("Dataset downloaded successfully!")
print("Dataset location:")
print(dataset_path)
```

```python
# ============================================================
# CELL 3: FIND IMAGE FILES AND FOLDERS
# ============================================================

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

print("BEFORE:")
print("Image folders have not been identified.")         #these are the files which are extracted

# Find all image files recursively
all_image_paths = []                        #we create an empty list

for root, dirs, files in os.walk(dataset_path):                           #it searches through the dataset folder by folder
    for file in files:                                                #roots-current folder,dirs-folder inside it,files-files inside it
            if file.lower().endswith(image_extensions):                #convert lower case
                        all_image_paths.append(os.path.join(root, file))

                        print("\nAFTER:")
                        print("Total image files found:", len(all_image_paths))

                        # Display first few image paths
                        print("\nFirst 5 image paths:")

                        for image_path in all_image_paths[:5]:
                            print(image_path)
```

```python
# ============================================================
# CELL 4: CLASS DISTRIBUTION
# ============================================================

from collections import Counter

print("BEFORE:")
print("Class distribution has not been calculated.")

# Take the parent folder name as the class name
class_names = [
    os.path.basename(os.path.dirname(path))
        for path in all_image_paths
        ]

class_counts = Counter(class_names)

print("\nAFTER:")
print("Classes found:")

for class_name, count in class_counts.items():
    print(f"{class_name}: {count} images")

print("\nTotal images:", len(all_image_paths))
```

```python
# ============================================================
# CELL 5: LOAD ONE RAW IMAGE
# ============================================================

print("BEFORE:")
print("No image has been loaded.")

# Select first image
sample_path = all_image_paths[0]

# Load image using OpenCV
raw_image_bgr = cv2.imread(sample_path)

# Check whether image was loaded
if raw_image_bgr is None:
    raise ValueError("Image could not be loaded.")

    print("\nAFTER:")
    print("Image loaded successfully!")
    print("Image file:", os.path.basename(sample_path))
    print("Image path:", sample_path)
    print("Shape:", raw_image_bgr.shape)
    print("Data type:", raw_image_bgr.dtype)
```

```python
# ============================================================
# CELL 6: DISPLAY RAW IMAGE
# ============================================================

print("BEFORE:")
print("Original RAW image:")

# Convert BGR to RGB for correct display
raw_image_rgb = cv2.cvtColor(raw_image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(raw_image_rgb)
plt.title("BEFORE - RAW IMAGE")
plt.axis("off")
plt.show()

print("\nAFTER:")
print("Raw image successfully displayed.")
```

```python
# ============================================================
# CELL 7: IMAGE PROPERTIES
# ============================================================

print("BEFORE:")
print("Image properties are not yet checked.")

height, width, channels = raw_image_rgb.shape

print("\nAFTER:")
print("Height:", height)
print("Width:", width)
print("Channels:", channels)
print("Minimum pixel:", raw_image_rgb.min())
print("Maximum pixel:", raw_image_rgb.max())
print("Data type:", raw_image_rgb.dtype)
```

```python
# ============================================================
# CELL 8: CORRUPTED IMAGE CHECK
# ============================================================

print("BEFORE:")
print("Dataset images have not been checked for corruption.")

valid_images = []
corrupted_images = []

for path in all_image_paths:
    try:
        # Open image using PIL
        img = Image.open(path)

        # Verify image integrity
        img.verify()

        valid_images.append(path)

    except Exception:
        corrupted_images.append(path)

print("\nAFTER:")
print("Image quality check completed.")

print("Total images    :", len(all_image_paths))
print("Valid images    :", len(valid_images))
print("Corrupted images:", len(corrupted_images))
```

```python
# ============================================================
# CELL 9: RESIZE IMAGE
# ============================================================

print("BEFORE:")
print("Original image size:", raw_image_rgb.shape[:2])

# Resize image to 224 × 224
resized_image = cv2.resize(
    raw_image_rgb,
        (224, 224),
            interpolation=cv2.INTER_AREA
            )

print("\nAFTER:")
print("Resized image size:", resized_image.shape[:2])

            # BEFORE and AFTER
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(raw_image_rgb)
plt.title(f"BEFORE - {width} × {height}")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(resized_image)
plt.title("AFTER - 224 × 224")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# CELL 10: BGR → RGB CONVERSION
# ============================================================

print("BEFORE:")
print("Image loaded by OpenCV is in BGR format.")

# Convert BGR → RGB
rgb_image = cv2.cvtColor(
    resized_image,
        cv2.COLOR_BGR2RGB
        )

print("\nAFTER:")
print("Image converted to RGB format.")
print("Shape:", rgb_image.shape)

        # Display before and after
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(resized_image)
plt.title("BEFORE - BGR data")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rgb_image)
plt.title("AFTER - RGB")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# CELL 11: NOISE REDUCTION
# ============================================================

print("BEFORE:")
print("Image before noise reduction.")

# Apply Gaussian Blur
denoised_image = cv2.GaussianBlur(
    rgb_image,
        (5, 5),
            0
            )

print("\nAFTER:")
print("Noise reduction completed.")
print("Gaussian Blur kernel: 5 × 5")

            # Display BEFORE and AFTER
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("BEFORE - Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(denoised_image)
plt.title("AFTER - Noise Reduced")
plt.axis("off")

plt.show()
```

```python




# ============================================================
# CELL 12: CONTRAST ENHANCEMENT
# ============================================================

print("BEFORE:")
print("Image before contrast enhancement.")

# Convert RGB → LAB
lab_image = cv2.cvtColor(
    denoised_image,
        cv2.COLOR_RGB2LAB
        )

        # Split LAB channels
L, A, B = cv2.split(lab_image)

        # Create CLAHE
clahe = cv2.createCLAHE(
            clipLimit=2.0,
                tileGridSize=(8, 8)
                )

                # Enhance brightness/contrast channel
L_enhanced = clahe.apply(L)

                # Merge channels
enhanced_lab = cv2.merge(
                    (L_enhanced, A, B)
                    )

                    # Convert LAB → RGB
enhanced_image = cv2.cvtColor(
                        enhanced_lab,
                            cv2.COLOR_LAB2RGB
                            )

print("\nAFTER:")
print("Contrast enhancement completed using CLAHE.")

                            # Display BEFORE and AFTER
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(denoised_image)
plt.title("BEFORE - Normal Contrast")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(enhanced_image)
plt.title("AFTER - CLAHE Enhanced")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# CELL 13: PIXEL NORMALIZATION
# ============================================================

print("BEFORE:")
print("Minimum pixel value:", enhanced_image.min())
print("Maximum pixel value:", enhanced_image.max())
print("Data type:", enhanced_image.dtype)

# Convert pixel values from 0–255 to 0–1
normalized_image = (
    enhanced_image.astype(np.float32) / 255.0
    )

print("\nAFTER:")
print("Minimum pixel value:", normalized_image.min())
print("Maximum pixel value:", normalized_image.max())
print("Data type:", normalized_image.dtype)

# Display BEFORE and AFTER
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(enhanced_image)
plt.title("BEFORE - Pixel Range 0–255")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(normalized_image)
plt.title("AFTER - Normalized 0–1")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# CELL 14: DATA AUGMENTATION
# ============================================================

print("BEFORE:")
print("Original preprocessed image.")

# Original image
original = normalized_image.copy()

# Horizontal flip
flipped = np.fliplr(original)

# Rotation
rotation_matrix = cv2.getRotationMatrix2D(
    (112, 112),
        15,
            1.0
            )

rotated = cv2.warpAffine(
                original,
                    rotation_matrix,
                        (224, 224)
                        )

                        # Zoom
zoomed = cv2.resize(
                            original[20:204, 20:204],
                                (224, 224)
                                )

print("\nAFTER:")
print("Data augmentation completed.")
print("Created:")
print("✓ Original")
print("✓ Horizontal flip")
print("✓ Rotation")
print("✓ Zoom")

                                # Display BEFORE and AFTER examples
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(original)
plt.title("BEFORE - Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(flipped)
plt.title("AFTER - Horizontal Flip")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rotated)
plt.title("AFTER - Rotation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(zoomed)
plt.title("AFTER - Zoom")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def preprocess_image(image_path):

    # 1. Read image
    image = cv2.imread(image_path)

    # 2. Convert BGR → RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # 3. Resize
    image = cv2.resize(image, (224, 224))

    # 4. Noise reduction
    image = cv2.GaussianBlur(image, (5, 5), 0)

    # 5. Contrast enhancement
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    L = clahe.apply(L)

    lab = cv2.merge((L, A, B))
    image = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)

    # 6. Normalize
    image = image.astype(np.float32) / 255.0

    return image
```

```python
# ============================================================
# CELL 17: BATCH PREPROCESSING
# ============================================================

import numpy as np
import os
from PIL import Image
import cv2

# Ensure valid_images is defined. This code is copied from cell -JcRODbE8JCP
# to handle potential kernel restarts without re-running all previous cells.
# It assumes 'all_image_paths' is defined from CELL 3 and 'Image' (from PIL) from CELL 1.
if 'valid_images' not in globals() or not isinstance(valid_images, list) or not valid_images:
    print("Re-validating images as 'valid_images' not found or empty...")
    valid_images = []
    corrupted_images = []
    if 'all_image_paths' in globals(): # Ensure all_image_paths is available
        for path in all_image_paths:
            try:
                img = Image.open(path)
                img.verify()
                valid_images.append(path)
            except Exception:
                corrupted_images.append(path)
        if corrupted_images:
            print(f"Warning: {len(corrupted_images)} images were found corrupted during re-validation.")
        else:
            print("'valid_images' re-validated successfully.")
    else:
        print("Error: 'all_image_paths' is not defined. Cannot re-validate images.")
        # If all_image_paths is not defined, valid_images will remain empty, which will cause issues.
        # For now, let's proceed assuming all_image_paths will be available in a complete run.

processed_images = []
processed_labels = []

print("BEFORE:")
print("Raw dataset images:", len(valid_images))
print("Starting image preprocessing...")

for image_path in valid_images:
    image = preprocess_image(image_path)
    processed_images.append(image)

    # Folder name = class label
    label = os.path.basename(
        os.path.dirname(image_path)
    )
    processed_labels.append(label)

processed_images = np.array(processed_images)

print("\nAFTER:")
print("Processed images:", len(processed_images))
print("Image shape:", processed_images.shape)
print("Pixel range:",
    processed_images.min(),
    "to",
    processed_images.max())
```
